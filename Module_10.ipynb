1. What is a random variable in probability theory?

A random variable is a value that comes from a random experiment.
It represents outcomes in numerical form.
For example, tossing a coin: Head = 1, Tail = 0.
It helps us study uncertainty using numbers.


2. What are the types of random variables?

There are two main types: Discrete and Continuous.
Discrete random variables take countable values (like 0, 1, 2).
Continuous random variables take infinite values within a range.
Example: Number of students (discrete), height (continuous).


3. What is the difference between discrete and continuous distributions?

Discrete distribution deals with countable values.
Continuous distribution deals with values in a continuous range.
Discrete uses probability mass function (PMF).
Continuous uses probability density function (PDF).


4. What are probability distribution functions (PDF)?

PDF shows how probability is distributed for a continuous variable.
It tells the likelihood of values in a range.
The total area under the PDF curve is 1.
It cannot give probability of one exact value.


5. How do CDF differ from PDF?

PDF gives probability density at a value.
CDF gives cumulative probability up to a value.
CDF is obtained by integrating the PDF.
CDF always increases from 0 to 1.


6. What is a discrete uniform distribution?

In this distribution, all values have equal probability.
Each outcome is equally likely.
Example: Rolling a fair dice.
Probability of each number is 1/6.


7. What are the key properties of a Bernoulli distribution?

It has only two outcomes: Success (1) and Failure (0).
Probability of success is p.
Probability of failure is 1 − p.
It is used for single yes/no experiments.


8. What is the binomial distribution?

It shows probability of getting success in fixed trials.
Each trial is independent.
Probability of success remains constant.
Example: Tossing a coin 5 times.


9. What is the Poisson distribution and where is it applied?

It shows number of events in a fixed time or area.
Events happen randomly and independently.
It uses a parameter called lambda (λ).
Example: Number of calls per hour.


10. What is a continuous uniform distribution?

All values in a given range are equally likely.
It is used for continuous data.
Example: Random number between 0 and 1.
Graph is a rectangle shape.


11. What are the characteristics of a normal distribution?

It is bell-shaped and symmetric.
Mean, median, and mode are equal.
Total area under curve is 1.
Most data lies near the mean.


12. What is the standard normal distribution?

It is a normal distribution with mean 0.
Its standard deviation is 1.
It is also called Z-distribution.
Used to calculate probabilities using Z-scores.


13. What is the Central Limit Theorem (CLT)?

CLT says sample means become normal when sample size is large.
It works even if population is not normal.
Usually sample size should be ≥ 30.
It is very important in statistics.


14. How does CLT relate to normal distribution?

CLT explains why many distributions look normal.
It says sampling distribution becomes normal.
As sample size increases, it becomes more accurate.
That is why normal distribution is widely used.


15. What is the application of Z statistics in hypothesis testing?

Z-statistics is used to test population mean.
It is used when population standard deviation is known.
It compares sample mean with population mean.
It helps decide to accept or reject hypothesis.


16. How do you calculate a Z-score?

Formula: Z = (X − μ) / σ
X is value, μ is mean, σ is standard deviation.
It shows how far a value is from mean.
It is measured in standard deviations.


17. What are point estimates and interval estimates?

Point estimate gives a single value estimate.
Example: sample mean.
Interval estimate gives a range of values.
Example: confidence interval.


18. What is the significance of confidence intervals?

It shows the range where true value may lie.
It gives more information than a single value.
Example: 95% confidence interval.
Higher confidence means wider interval.


19. What is relationship between Z-score and confidence interval?

Z-score is used to calculate confidence interval.
It depends on confidence level (like 95%).
Higher Z means wider interval.
It helps measure margin of error.


20. How are Z-scores used to compare distributions?

Z-scores standardize different data.
They remove unit differences.
We can compare values from different datasets.
Higher Z means value is above mean.


21. What are assumptions for applying CLT?

Samples must be random.
Observations must be independent.
Sample size should be large (≥30).
Population should not be extremely skewed.


22. What is expected value in probability?

Expected value is the average outcome.
It is calculated using probabilities.
It represents long-term mean.
Also called mean of distribution.


23. How does probability distribution relate to expected outcome?

Probability distribution shows all possible values.
Each value has its probability.
Expected value is calculated from this distribution.
It tells average result in long run.
