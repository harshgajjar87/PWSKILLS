1. What is hypothesis testing in statistics?

Hypothesis testing is a method used to make decisions using data.
It checks whether a claim about a population is true or not.
We compare sample data with a stated assumption.
Based on results, we accept or reject the hypothesis.

2. What is the null hypothesis, and how does it differ from the alternative hypothesis?

The null hypothesis (H₀) states there is no effect or no difference.
The alternative hypothesis (H₁) states there is an effect or difference.
H₀ is assumed true at the start.
We test data to see if H₀ should be rejected.

3. What is the significance level and why is it important?

The significance level is denoted by α (alpha).
Common values are 0.05 or 0.01.
It represents the risk of rejecting a true null hypothesis.
It helps control decision errors in testing.

4. What does a P-value represent in hypothesis testing?

The P-value shows how strong the evidence is against H₀.
It gives the probability of getting the observed result by chance.
A small P-value means strong evidence against H₀.
It helps in decision making.

5. How do you interpret the P-value?

If P-value ≤ α, reject the null hypothesis.
If P-value > α, fail to reject the null hypothesis.
Smaller P-value means more significant result.
It does not measure truth, only evidence.

6. What are Type I and Type II errors?

Type I error occurs when a true H₀ is rejected.
It is also called a false positive.
Type II error occurs when a false H₀ is accepted.
It is also called a false negative.

7. Difference between one-tailed and two-tailed tests?

One-tailed test checks effect in one direction only.
Two-tailed test checks effect in both directions.
One-tailed is more powerful in one direction.
Two-tailed is more commonly used.

8. What is the Z-test and when is it used?

Z-test is used to test population mean.
It is used when population standard deviation is known.
Sample size should be large (n ≥ 30).
It follows the normal distribution.

9. How do you calculate the Z-score and what does it represent?

Z-score is calculated as (X − μ) / σ.
It shows how far a value is from the mean.
Measured in standard deviation units.
Used to compare values across datasets.

10. What is the T-distribution and when should it be used?

T-distribution is similar to normal distribution.
It is used when sample size is small.
Population standard deviation is unknown.
It has thicker tails than normal distribution.

11. Difference between Z-test and T-test?

Z-test uses population standard deviation.
T-test uses sample standard deviation.
Z-test is for large samples.
T-test is for small samples.

12. What is the T-test and how is it used?

T-test compares sample means.
It checks if difference is statistically significant.
Used when population variance is unknown.
Common in small sample analysis.

13. Relationship between Z-test and T-test?

Both are hypothesis testing tools.
They test population means.
As sample size increases, T-test approaches Z-test.
Choice depends on sample size and variance.

14. What is a confidence interval?

A confidence interval gives a range of values.
It likely contains the true population parameter.
Common levels are 90%, 95%, and 99%.
It gives more information than point estimates.

15. What is margin of error and its effect?

Margin of error shows possible error in estimate.
It depends on confidence level and variability.
Larger margin gives wider confidence interval.
Smaller margin gives more precision.

16. How is Bayes’ Theorem used and why is it important?

Bayes’ theorem updates probability using new data.
It combines prior and current information.
Used in machine learning and decision making.
It improves prediction accuracy.

17. What is the Chi-square distribution and when is it used?

Chi-square distribution is a right-skewed distribution.
It is used for categorical data.
Commonly used in independence testing.
Also used in goodness of fit tests.

18. What is the Chi-square goodness of fit test?

It checks how well data fits a distribution.
Compares observed and expected frequencies.
Used for categorical variables.
Helps verify assumptions about data.

19. What is the F-distribution and when is it used?

F-distribution compares variances.
It is always positive and right-skewed.
Used in ANOVA and F-tests.
Depends on two degrees of freedom.

20. What is ANOVA and its assumptions?

ANOVA compares means of more than two groups.
It uses variance to test differences.
Assumes normality and equal variances.
Observations must be independent.

21. What are the different types of ANOVA tests?

One-way ANOVA compares one factor.
Two-way ANOVA compares two factors.
Repeated measures ANOVA uses same subjects.
MANOVA handles multiple dependent variables.

22. What is the F-test and how does it relate to hypothesis testing?

F-test compares variances or group means.
Used in ANOVA testing.
It checks overall significance of a model.
Decision is based on F-statistic value.

```python
#5 Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def hypothesis_test_with_errors(mu0=50, mu1=55, sigma=10, n=30, alpha=0.05):

    # Standard error
    se = sigma / np.sqrt(n)

    # Critical Z value (two-tailed)
    z_critical = norm.ppf(1 - alpha/2)

    # Convert to sample mean scale
    lower = mu0 - z_critical * se
    upper = mu0 + z_critical * se

    # Type II error (using SAME rejection region)
    beta = norm.cdf(upper, mu1, se) - norm.cdf(lower, mu1, se)

    # X values
    x = np.linspace(mu0 - 4*se, mu1 + 4*se, 1000)

    # Distributions
    y0 = norm.pdf(x, mu0, se)  # under H0
    y1 = norm.pdf(x, mu1, se)  # under H1

    plt.figure()
    plt.plot(x, y0)
    plt.plot(x, y1)

    # Type I error region
    plt.fill_between(x, y0, where=(x <= lower))
    plt.fill_between(x, y0, where=(x >= upper))

    # Type II error region
    plt.fill_between(x, y1, where=(x >= lower) & (x <= upper))

    plt.axvline(lower)
    plt.axvline(upper)

    plt.title("Type I and Type II Errors During Hypothesis Testing")
    plt.xlabel("Sample Mean")
    plt.ylabel("Probability Density")
    plt.show()

    print("Significance Level (Type I Error):", alpha)
    print("Type II Error:", beta)
    print("Power:", 1 - beta)
hypothesis_test_with_errors()
```

```python
#9 Write a Python function to calculate the confidence interval for a sample mean and explain its significance
import numpy as np
from scipy import stats

def confidence_interval(data, confidence=0.95, sigma=None):

    n = len(data)
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)

    alpha = 1 - confidence

    if sigma is not None:
        critical_value = stats.norm.ppf(1 - alpha/2)
        margin_error = critical_value * (sigma / np.sqrt(n))
        method = "Z-distribution"
    else:
        critical_value = stats.t.ppf(1 - alpha/2, df=n-1)
        margin_error = critical_value * (sample_std / np.sqrt(n))
        method = "T-distribution"

    lower = sample_mean - margin_error
    upper = sample_mean + margin_error

    print("Method Used:", method)
    print("Sample Mean:", sample_mean)
    print(f"{int(confidence*100)}% Confidence Interval: ({lower}, {upper})")

    return lower, upper
data = np.random.normal(23,12,1000)
confidence_interval(data, sigma=14)
```

```python
#10  Write a Python program to calculate the margin of error for a given confidence level using sample data
import numpy as np
from scipy import stats

def margin_of_error(data, confidence=0.95):

    n = len(data)
    sample_std = np.std(data, ddof=1)
    alpha = 1 - confidence

    t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

    standard_error = sample_std / np.sqrt(n)
    ME = t_critical * standard_error

    print("Sample Size:", n)
    print("Sample Standard Deviation:", sample_std)
    print("Confidence Level:", confidence)
    print("Margin of Error:", ME)

    return ME
data = np.random.normal(50, 10, 25)

margin_of_error(data, confidence=0.95)
```

```python
#19
import numpy as np
from scipy.stats import shapiro, levene

def check_anova_assumptions(groups):
    # Normality test (Shapiro-Wilk)
    for i, group in enumerate(groups):
        stat, p = shapiro(group)
        print(f"Group {i+1} Normality p-value:", p)

    # Equal variance test (Levene)
    stat, p = levene(*groups)
    print("Equal variance test p-value:", p)

    print("Independence must be ensured during data collection.")
g1 = np.random.normal(10,2,30)
g2 = np.random.normal(12,2,30)
g3 = np.random.normal(11,2,30)

check_anova_assumptions([g1,g2,g3])
```

```python
#24
import numpy as np
from scipy.stats import chi2

data = np.random.normal(10,2,30)

n = len(data)
sample_var = np.var(data, ddof=1)

sigma0 = 4   # hypothesized variance

chi_stat = (n-1)*sample_var/sigma0

p_value = 1-chi2.cdf(chi_stat, n-1)

print("Chi-square:",chi_stat)
print("p-value:",p_value)
```

```python
#25
from statsmodels.stats.proportion import proportions_ztest

success = [40, 30]
total = [100, 100]

stat,p = proportions_ztest(success,total)

print("Z-statistic:",stat)
print("p-value:",p)
```

```python
#26
import numpy as np
from scipy.stats import f
import matplotlib.pyplot as plt

data1 = np.random.normal(10,2,30)
data2 = np.random.normal(10,3,30)

var1 = np.var(data1,ddof=1)
var2 = np.var(data2,ddof=1)

F = var1/var2

print("F statistic:",F)

plt.boxplot([data1,data2],labels=["Dataset1","Dataset2"])
plt.title("Variance Comparison")
plt.show()
```

```python
#27
import numpy as np
from scipy.stats import chisquare

# Generate observed data
observed = np.random.randint(10,20,5)

# Calculate expected frequencies with same total
expected = np.ones(5) * (sum(observed)/5)

# Chi-square test
stat, p = chisquare(observed, expected)

print("Observed:", observed)
print("Expected:", expected)
print("Chi-square:", stat)
print("p-value:", p)
```
